## Text preprocessing
#### Steps:
- Text to lowercase
- Removing punctuations
- Tokenization
- Removal of stop words
- Lemmatization

### Text to lowercase

```python
# Text to lowercase function
def text_lowercase(text):
    return text.lower()

# Text to lowercase
corpus_02 = text_lowercase(corpus_01)
corpus_02
```

### Removing punctuation

```python
# Remove punctuation function
def remove_punctuation(text):
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)

# Remove punctuation
corpus_03 = remove_punctuation(corpus_02)
corpus_03
```

### Tokenize text & removal of stopwords

```python
# Show english stopwords
eng_stopwords = set(stopwords.words('english'))
print("List of english stopwords:")
print(eng_stopwords)
```

```python
# Function for tokenization and the removal of stopwords
def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word not in stop_words]
    return filtered_text
 
# Remove stopwords
corpus_04 = remove_stopwords(corpus_03)
print(corpus_04, end="")
```

### Lemmatization

```python
# Initialize Lemmatizer
lemmatizer = WordNetLemmatizer()

# Lemmatize string function
def lemmatize_word(text):
    word_tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(word, pos ='v') for word in word_tokens]
    return lemmas

# Lemmatize
lem = []
for i in corpus_04:
    lem.append(lemmatize_word(i))

# Nested list to list
corpus_05 = [' '.join([str(x) for x in lst]) for lst in lem]

print('Before lemmatization:')
print(corpus_04, '\n')

print('After lemmatization:')
print(corpus_05, end="")
```

## Term frequency-inverse document frequency (TF-IDF)
- For details see: https://www.learndatasci.com/glossary/tf-idf-term-frequency-inverse-document-frequency

### Term Frequency (TF)

```python
# Compute Term Frequency (TF)
words_set = set()
for doc in corpus:
    words = doc.split(' ')
    words_set = words_set.union(set(words))
    
print('Number of words in the corpus:',len(words_set), '\n')
print('The words in the corpus: \n', words_set)

# Number of documents in the corpus
n_docs = len(corpus)

# Number of unique words in the corpus 
n_words_set = len(words_set)

df_tf = pd.DataFrame(np.zeros((n_docs, n_words_set)), 
                     columns=list(words_set))

print("\nTerm Frequency (TF):")
for i in range(n_docs):
    # Words in the document
    words = corpus[i].split(' ')
    for w in words:
        df_tf[w][i] = df_tf[w][i] + (1 / len(words))
        
print(df_tf.round(4))
```

### Inverse Document Frequency (IDF)

```python
# Computing Inverse Document Frequency (IDF)
print("\nInverse Document Frequency (IDF):")

idf = {}

for w in words_set:
    
    # k = number of documents that contain this word
    k = 0
    
    for i in range(n_docs):
        if w in corpus[i].split():
            k += 1
            
    idf[w] =  np.log10(n_docs / k).round(4)
    
    print(f'{w:>15}: {idf[w]:>10}')
```

### Term Frequency - Inverse Document Frequency (TF-IDF)

```python
# Computing TF-IDF
df_tf_idf = df_tf.copy()

for w in words_set:
    for i in range(n_docs):
        df_tf_idf[w][i] = df_tf[w][i] * idf[w]

print('\nTF-IDF:')
print(df_tf_idf.round(4))
```

## Part-of-Speach (POS) tagging
For meaning of POS-tags see: https://pythonexamples.org/nltk-pos-tagging

```python
text = '''European authorities fined Google a record $5.1 
          billion on Wednesday for abusing its power in the 
          mobile phone market.'''

def preprocess(sent):
    sent = nltk.word_tokenize(sent)
    sent = nltk.pos_tag(sent)
    return sent

sent = preprocess(text)
pattern = 'NP: {<DT>?<JJ>*<NN>}'

cp = nltk.RegexpParser(pattern)
cs = cp.parse(sent)

iob_tagged = tree2conlltags(cs)

# Print the POS-tags
pprint(iob_tagged)
```

### Explanation of POS Tags

Based on the custom text example above, here is a detailed explanation of at least 5 POS tags found in the output:

**1. NN (Noun, Singular)**
   - Definition: A singular noun (a person, place, thing, or idea)
   - Examples from output: 'learning', 'algorithms', 'patterns', 'data', 'networks', 'results', 'tasks'
   - Usage: Used for single, countable nouns

**2. VB (Verb, Base Form)**
   - Definition: A verb in its base form (infinitive), without any tense or person modification
   - Examples from output: 'learn', 'achieve'
   - Usage: Root form of verbs, often used after modals like 'can', 'will', etc.

**3. VBP (Verb, Non-3rd Person Singular Present)**
   - Definition: A present tense verb for non-3rd person subjects (I, you, we, they)
   - Examples from output: 'learn' (in "algorithms learn")
   - Usage: Indicates actions occurring now for plural or non-third person subjects

**4. JJ (Adjective)**
   - Definition: A word that describes or modifies a noun
   - Examples from output: 'complex', 'excellent', 'powerful', 'Machine'
   - Usage: Provides additional information about qualities or characteristics of nouns

**5. NNS (Noun, Plural)**
   - Definition: A plural noun (multiple persons, places, things, or ideas)
   - Examples from output: 'patterns', 'algorithms', 'networks', 'results', 'tasks'
   - Usage: Used for nouns representing more than one entity

**Additional Tags Explained:**

**6. DT (Determiner)**
   - Definition: Words that introduce or specify a noun
   - Examples from output: 'The'
   - Usage: Articles and other determiners that come before nouns

**7. IN (Preposition or Subordinating Conjunction)**
   - Definition: Words that show relationships between nouns/pronouns and other words
   - Examples from output: 'from', 'on'
   - Usage: Indicates spatial, temporal, or logical relationships

**8. PRP (Personal Pronoun)**
   - Definition: Pronouns that refer to persons or things
   - Examples from output: Would appear if pronouns like 'he', 'she', 'it', 'they' were present
   - Usage: Replaces nouns to avoid repetition
